```python
import pandas as pd

df = pd.read_csv("../datasets/raw/Modified_SQL_Dataset.csv")

# 1. Tổng số dòng
print("Tổng số dòng:", len(df))

# 2. Số dòng trùng lặp
n_dup = df.duplicated().sum()
print("Số dòng trùng lặp:", n_dup)
print("Số dòng sau khi loại trùng:", len(df) - n_dup)

# 3. Số dòng train/test sau chia (chạy sau khi đã có train_df, test_df)
# Nếu đã lưu file rồi thì đọc lại:
train_df = pd.read_csv("../datasets/processed/train.csv")
test_df = pd.read_csv("../datasets/processed/test.csv")
print("Số dòng train:", len(train_df))
print("Số dòng test:", len(test_df))

# 4. Tỷ lệ nhãn tổng thể (trên dataset đã loại trùng, trước khi chia)
df_clean = df.drop_duplicates(subset=['Query']).reset_index(drop=True)
print("Tỷ lệ nhãn (toàn bộ dataset):\n", df_clean['Label'].value_counts(normalize=True))
print("Tỷ lệ nhãn (train):\n", train_df['Label'].value_counts(normalize=True))
print("Tỷ lệ nhãn (test):\n", test_df['Label'].value_counts(normalize=True))

# 5. Kiểm tra null
print("Giá trị null theo cột:\n", df.isnull().sum())
```
